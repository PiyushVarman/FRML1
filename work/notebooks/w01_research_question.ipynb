## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Direct raw URL to the CSV in your GitHub repository
url = "https://raw.githubusercontent.com/PiyushVarman/FRML1/main/data/raw/content_refresh_anonymized.csv"

# Load the dataset directly into pandas
df = pd.read_csv(url)

# Apply standard starter filter: mature content (age >= 90d) with search visibility (impressions > 0)
df_mature = df[(df["content_age_days"] >= 90) & (df["impressions_90d"] > 0)].copy()
total_mature = len(df_mature)

# 1. Decay prevalence
declining_count = (df_mature["trend_direction"] == "down").sum()
declining_pct = (declining_count / total_mature) * 100

# 2. Stale high-visibility inventory
stale_visible = df_mature[
    (df_mature["days_since_last_update"] >= 180) & (df_mature["impressions_90d"] >= 500)
]
stale_visible_count = len(stale_visible)

# 3. Striking distance with low CTR opportunity
striking_low_ctr = df_mature[
    (df_mature["impressions_90d"] >= 500)
    & (df_mature["avg_position"] > 3)
    & (df_mature["avg_position"] <= 20)
    & (df_mature["ctr"] < 0.5)
]
striking_low_ctr_count = len(striking_low_ctr)

print(f"Total Mature Pages Analyzed: {total_mature:,}")
print("-" * 65)
print(f"1. Declining Trend Rate:        {declining_count:,} pages ({declining_pct:.2f}%)")
print(f"2. Stale Visible Inventory:     {stale_visible_count:,} pages (>=500 imps, >=180d untouched)")
print(f"3. Striking-Distance Low CTR:   {striking_low_ctr_count:,} pages (pos 4-20, imps >=500, ctr <0.5%)")
```
